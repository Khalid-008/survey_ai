```python
import plotly.graph_objects as go

months = [
    "2024-04", "2024-05", "2024-06", "2024-07", "2024-08", "2024-09",
    "2024-10", "2024-11", "2024-12", "2025-01", "2025-02", "2025-03",
    "2025-04", "2025-05", "2025-06", "2025-07", "2025-08", "2025-09",
    "2025-10", "2025-11", "2025-12", "2026-01", "2026-02", "2026-03"
]

avg_scores = [
    5.00, 4.85, 4.75, 4.64, 4.51, 4.76,
    4.76, 4.90, 4.64, 4.66, 4.29, 4.86,
    4.83, 4.84, 4.91, 4.91, 4.80, 4.65,
    4.79, 4.93, 4.78, 4.92, 4.83, 4.87
]

# Color each point based on value
colors = ["#ef4444" if s < 4.6 else "#f59e0b" if s < 4.75 else "#22c55e" for s in avg_scores]

fig = go.Figure()

# Shaded area under the line
fig.add_trace(go.Scatter(
    x=months,
    y=avg_scores,
    fill="tozeroy",
    fillcolor="rgba(99, 102, 241, 0.08)",
    line=dict(color="rgba(0,0,0,0)"),
    showlegend=False,
    hoverinfo="skip"
))

# Main line
fig.add_trace(go.Scatter(
    x=months,
    y=avg_scores,
    mode="lines+markers",
    name="Avg Score",
    line=dict(color="#6366f1", width=2.5, shape="spline", smoothing=0.4),
    marker=dict(
        size=8,
        color=colors,
        line=dict(color="white", width=1.5)
    ),
    hovertemplate="<b>%{x}</b><br>Score: %{y:.2f}<extra></extra>"
))

# Reference line at 4.75
fig.add_hline(
    y=4.75,
    line_dash="dot",
    line_color="rgba(156,163,175,0.6)",
    line_width=1.5,
    annotation_text="4.75 baseline",
    annotation_position="right",
    annotation_font=dict(size=11, color="#9ca3af")
)

# Annotate min and max
min_val = min(avg_scores)
max_val = max(avg_scores)
min_month = months[avg_scores.index(min_val)]
max_month = months[avg_scores.index(max_val)]

fig.add_annotation(
    x=min_month, y=min_val,
    text=f"Min: {min_val}",
    showarrow=True, arrowhead=2, arrowcolor="#ef4444",
    font=dict(color="#ef4444", size=11),
    ax=0, ay=30
)
fig.add_annotation(
    x=max_month, y=max_val,
    text=f"Max: {max_val}",
    showarrow=True, arrowhead=2, arrowcolor="#22c55e",
    font=dict(color="#22c55e", size=11),
    ax=0, ay=-35
)

fig.update_layout(
    title=dict(
        text="Monthly Average Score",
        font=dict(size=20, color="#1e1b4b"),
        x=0.05
    ),
    xaxis=dict(
        title="Month",
        tickangle=-45,
        tickfont=dict(size=11),
        gridcolor="rgba(229,231,235,0.5)",
        showline=True,
        linecolor="#e5e7eb"
    ),
    yaxis=dict(
        title="Average Score",
        range=[4.1, 5.15],
        tickfont=dict(size=11),
        gridcolor="rgba(229,231,235,0.5)",
        showline=True,
        linecolor="#e5e7eb",
        dtick=0.1
    ),
    plot_bgcolor="white",
    paper_bgcolor="white",
    hovermode="x unified",
    showlegend=False,
    margin=dict(l=60, r=80, t=70, b=80),
    width=1000,
    height=480
)

fig.show()
```
